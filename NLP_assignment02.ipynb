```python
from nltk import CFG
from nltk.parse import ChartParser

grammar = CFG.fromstring("""
S -> Decl | Imp | YesNoQ
Decl -> NP VP
Imp -> V NP
YesNoQ -> Aux NP VP
NP -> Det N
VP -> V NP
Det -> 'the'
N -> 'student' | 'form'
V -> 'submits' | 'check'
Aux -> 'does'
""")

parser = ChartParser(grammar)

sentences = [
    "the student submits the form",
    "check the form"
]

for sentence in sentences:
    print("Sentence:", sentence)
    for tree in parser.parse(sentence.split()):
        print(tree)
```

```python
def check_agreement(sentence):
    valid_sentences = [
        "the student submits the form",
        "the students submit the form",
        "she checks the timetable",
        "they check the timetable"
    ]

    if sentence.lower() in valid_sentences:
        print("Grammatical Sentence")
    else:
        print("Ungrammatical Sentence")


check_agreement("the student submits the form")
check_agreement("the students submits the form")
```

```python
verb_frames = {
    "crashed": "Intransitive",
    "submit": "Transitive",
    "send": "Ditransitive",
    "tell": "Sentential Complement",
    "apply": "PP Complement"
}

for verb, frame in verb_frames.items():
    print(verb, ":", frame)

sentence = "submit the form"

if "submit" in sentence and "form" in sentence:
    print("Valid Verb Structure")
else:
    print("Invalid Verb Structure")
```

```python
def unify(feature1, feature2):
    result = feature1.copy()

    for key, value in feature2.items():
        if key in result and result[key] != value:
            return "Unification Failed"

        result[key] = value

    return result


f1 = {"NUM": "singular", "PER": 3}
f2 = {"NUM": "singular"}

f3 = {"CASE": "nominative"}
f4 = {"CASE": "accusative"}

print(unify(f1, f2))
print(unify(f3, f4))
```

```python
!pip install nltk

import time
from nltk import CFG
from nltk.parse import (
    RecursiveDescentParser,
    EarleyChartParser
)

grammar = CFG.fromstring("""
S -> NP VP
NP -> Det N | Pron | NP PP
VP -> Vt NP | Vi | VP PP
PP -> P NP
Det -> 'the' | 'a'
N -> 'student' | 'students' | 'timetable' | 'form' | 'library'
Pron -> 'they' | 'she'
Vt -> 'checks' | 'submit' | 'show'
Vi -> 'arrive' | 'crash'
P -> 'near' | 'with'
""")

tests = [
    "the student checks the timetable",
    "they show the form",
    "the students arrive",
    "the student checks the timetable near the library"
]

parsers = [
    ("Top Down", RecursiveDescentParser(grammar)),
    ("Earley", EarleyChartParser(grammar))
]

for sentence in tests:
    print("\\nSentence:", sentence)

    for name, parser in parsers:
        start = time.perf_counter()
        trees = list(parser.parse(sentence.split()))
        elapsed = time.perf_counter() - start

        print(name)
        print("Number of Parses:", len(trees))
        print("Execution Time:", round(elapsed, 6), "seconds")
```

```python
from nltk import CFG
from nltk.parse import ChartParser

grammar = CFG.fromstring("""
S -> NP VP | S Conj S
NP -> Det N | Pron
VP -> V NP | V
Det -> 'the' | 'my'
N -> 'student' | 'timetable' | 'fee'
Pron -> 'I' | 'you'
V -> 'need' | 'help'
Conj -> 'and'
""")

parser = ChartParser(grammar)

sentences = [
    "I need my timetable",
    "the student need the fee",
    "I need my timetable and you help"
]

for sentence in sentences:
    print("\nInput:", sentence)

    trees = list(parser.parse(sentence.split()))

    if trees:
        for tree in trees:
            print(tree)
    else:
        print("No complete parse found")
```
