```python
def generate_word(word, rule):
    if rule == "PL":
        if word.endswith("y") and len(word) > 1:
            if word[-2] not in "aeiou":
                return word[:-1] + "ies"
        if word.endswith(("s", "sh", "ch", "x", "z")):
            return word + "es"
        return word + "s"

    elif rule == "3SG":
        if word.endswith("y") and len(word) > 1:
            if word[-2] not in "aeiou":
                return word[:-1] + "ies"
        if word.endswith(("s", "sh", "ch", "x", "z", "o")):
            return word + "es"
        return word + "s"

    elif rule == "PROG":
        if word.endswith("ie"):
            return word[:-2] + "ying"
        if word.endswith("e") and not word.endswith("ee"):
            return word[:-1] + "ing"
        return word + "ing"

    elif rule == "PAST":
        if word.endswith("e"):
            return word + "d"
        if word.endswith("y") and len(word) > 1:
            if word[-2] not in "aeiou":
                return word[:-1] + "ied"
        return word + "ed"

    else:
        return "Invalid rule"


word = input("Enter a word: ").lower()
rule = input("Enter a rule (PL, 3SG, PROG, PAST): ").upper()

result = generate_word(word, rule)
print("Generated word:", result)
```

```python
lexicon = {
    "cat": "N",
    "dog": "N",
    "walk": "V",
    "play": "V"
}

word = input("Enter a word: ").lower()

if word in lexicon:
    print(word, "+", lexicon[word])

elif word.endswith("s") and word[:-1] in lexicon:
    print(word[:-1], "+", lexicon[word[:-1]], "+ PL")

elif word.endswith("ed") and word[:-2] in lexicon:
    print(word[:-2], "+", lexicon[word[:-2]], "+ PAST")

else:
    print("UNKNOWN")
```
