```python

words = ["connected", "connecting", "connection"]

print(f"{'Word':<15}{'Root':<12}{'Suffix':<12}{'Type':<18}{'Normalized'}")
print("-" * 70)

for word in words:
    if word == "connected":
        root, suffix, kind, base = "connect", "ed", "Inflectional", "connect"
    elif word == "connecting":
        root, suffix, kind, base = "connect", "ing", "Inflectional", "connect"
    elif word == "connection":
        root, suffix, kind, base = "connect", "ion", "Derivational", "connect"

    print(f"{word:<15}{root:<12}{suffix:<12}{kind:<18}{base}")
```

```python

words = ["unhappy", "happiness", "happily"]

print(f"{'Word':<12}{'Prefix':<10}{'Root':<10}{'Suffix':<10}{'Type':<15}{'Normalized'}")
print("-" * 75)

for word in words:
    if word == "unhappy":
        prefix, root, suffix, kind = "un", "happy", "-", "Derivational"
    elif word == "happiness":
        prefix, root, suffix, kind = "-", "happy", "ness", "Derivational"
    else:
        prefix, root, suffix, kind = "-", "happy", "ly", "Derivational"

    print(f"{word:<12}{prefix:<10}{root:<10}{suffix:<10}{kind:<15}{root}")
```

```python

words = ["played", "player", "playing"]

print(f"{'Word':<12}{'Stem':<12}{'Affix':<10}{'Type':<15}{'Normalized'}")
print("-" * 65)

for word in words:
    if word == "played":
        stem, affix, kind = "play", "ed", "Inflectional"
    elif word == "player":
        stem, affix, kind = "play", "er", "Derivational"
    else:
        stem, affix, kind = "play", "ing", "Inflectional"

    print(f"{word:<12}{stem:<12}{affix:<10}{kind:<15}{stem}")
```

```python

words = ["writes", "writing", "written"]

rules = {
    "writes":  ("write", "s", "Regular", "q0 -> q1 -> q2 -> qf"),
    "writing": ("write", "ing", "Regular", "q0 -> q1 -> q3 -> qf"),
    "written": ("write", "ten", "Irregular", "q0 -> q1 -> q4 -> qf")
}

print(f"{'Word':<12}{'Root':<10}{'Component':<12}{'Class':<12}State Path")
print("-" * 80)

for word in words:
    root, component, kind, path = rules[word]
    print(f"{word:<12}{root:<10}{component:<12}{kind:<12}{path}")
```

```python

from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
words = ["relational", "relation", "relate"]

print(f"{'Word':<15}{'Step 1':<15}{'Final Stem':<15}")
print("-" * 45)

for word in words:
    if word == "relational":
        step = "relate"
    elif word == "relation":
        step = "relat"
    else:
        step = "relate"

    stem = stemmer.stem(word)
    print(f"{word:<15}{step:<15}{stem:<15}")
```
