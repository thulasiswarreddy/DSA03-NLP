```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from collections import Counter

nltk.download('punkt')

paragraph = input("Enter a paragraph: ")

# a) Split into sentences
sentences = sent_tokenize(paragraph)
print("\nSentences:")
for s in sentences:
    print(s)

# b) Split into words
words = word_tokenize(paragraph)
print("\nWords:")
print(words)

# c) Count the words
count = Counter(words)
print("\nWord Count:")
for word, c in count.items():
    print(word, ":", c)

# d) Most frequent words
print("\nMost Frequent Words:")
for word, c in count.most_common(3):
    print(word, ":", c)
```

```python
from collections import Counter


text = """
the cat is in the park
the man is in the park
the cat and the man are friends
the book is on the table
"""


words = text.lower().split()
word_counts = Counter(words)


def edits1(word):
    letters = 'abcdefghijklmnopqrstuvwxyz'
    splits = [(word[:i], word[i:]) for i in range(len(word) + 1)]

    deletes = [L + R[1:] for L, R in splits if R]
    inserts = [L + c + R for L, R in splits for c in letters]
    replaces = [L + c + R[1:] for L, R in splits if R
                for c in letters]
    transposes = [L + R[1] + R[0] + R[2:]
                  for L, R in splits if len(R) > 1]

    return set(deletes + inserts + replaces + transposes)


def correct(word):
    word = word.lower()


    if word in word_counts:
        return word

    candidates = edits1(word)
    known = [w for w in candidates if w in word_counts]

    if known:
        return max(known, key=lambda w: word_counts[w])

    return word


test_words = ["ten", "cta", "mann", "boook", "park"]

print("Word\tCorrected Word")
for word in test_words:
    print(word, "\t", correct(word))
```

```python
import re

text = input("Enter the text: ")


date_pattern = r'\b\d{1,2}[/-]\d{1,2}[/-]\d{4}\b'
first_date = re.search(date_pattern, text)


mobile_pattern = r'\b[6-9]\d{9}\b'
mobile_numbers = re.findall(mobile_pattern, text)


email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'
emails = re.findall(email_pattern, text)


capital_pattern = r'\b[A-Z][a-z]*\b'
capital_words = re.findall(capital_pattern, text)


starts_with_call = bool(re.match(r'(?i)^call', text))


changed_text = re.sub(
    r'\b(\d{2})/(\d{2})/(\d{4})\b',
    r'\3-\2-\1',
    text
)


hidden_text = re.sub(
    r'\b[6-9]\d{9}\b',
    'XXXXXXXXXX',
    changed_text
)

print("\nFirst Date:", first_date.group() if first_date else "Not found")
print("Mobile Numbers:", mobile_numbers)
print("Email IDs:", emails)
print("Capitalized Words:", capital_words)
print('Starts with "call":', starts_with_call)
print("Changed Date Text:", changed_text)
print("Hidden Phone Text:", hidden_text)
```

```python
import re

patterns = {
    "a*": r"a*",
    "a+": r"a+",
    "a?": r"a?",
    "[abc]": r"[abc]",
    "[^abc]": r"[^abc]",
    "^abc": r"^abc",
    "abc$": r"abc$"
}

strings = ["abc", "aaa", "bbb", "ab", "cab", "xyz", "abc123"]

for pattern_name, pattern in patterns.items():
    print("\nPattern:", pattern_name)

    for string in strings:
        if re.fullmatch(pattern, string):
            print(string, "YES")
        else:
            print(string, "NO")
```

```python


def fsa(string):
    state = "q0"
    path = [state]

    for ch in string:
        if state == "q0":
            if ch == "a":
                state = "q1"
            else:
                state = "q0"

        elif state == "q1":
            if ch == "a":
                state = "q1"
            else:
                state = "q2"

        elif state == "q2":
            if ch == "a":
                state = "q1"
            else:
                state = "q0"

        path.append(state)

    print("Path:", " -> ".join(path))

    if state == "q2":
        print("ACCEPTED")
    else:
        print("REJECTED")



string = input("Enter a string using a and b: ")

fsa(string)
```

```python
import re

text = "My phone number is 9876543210 and my email is anu@gmail.com."


phone = re.search(r'\b[6-9]\d{9}\b', text)


email = re.search(r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b', text)


word = re.match(r'My', text)


if phone:
    print("Mobile Number Found:", phone.group())
else:
    print("Mobile Number Not Found")

if email:
    print("Email Found:", email.group())
else:
    print("Email Not Found")

if word:
    print("Pattern matched at beginning: YES")
else:
    print("Pattern matched at beginning: NO")
```

```python

def automaton(string):
    state = "q0"
    path = [state]

    for ch in string:
        if state == "q0":
            if ch == "a":
                state = "q1"
            elif ch == "b":
                state = "q0"

        elif state == "q1":
            if ch == "a":
                state = "q1"
            elif ch == "b":
                state = "q2"

        elif state == "q2":
            if ch == "a":
                state = "q1"
            elif ch == "b":
                state = "q0"

        path.append(state)

    print("State Path:", " -> ".join(path))

    if state == "q2":
        print("ACCEPTED")
    else:
        print("REJECTED")


string = input("Enter a string containing a and b: ")

automaton(string)
```
