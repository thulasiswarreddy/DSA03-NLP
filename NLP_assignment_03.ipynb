```python
import re

text = """
Patient ID: P12345
Date: 12-10-2026
Drug: Paracetamol 500mg
Phone: 9876543210
"""

patterns = {
    "Patient ID": r'P\d{5}',
    "Date": r'\d{2}-\d{2}-\d{4}',
    "Drug Strength": r'[A-Za-z]+\s\d+\s?mg',
    "Phone": r'\b\d{10}\b'
}

for name, pattern in patterns.items():
    result = re.findall(pattern, text)
    print(name, ":", result)
```

```python
import re

text = "Patient has FEVER and BP 120/80. Given PCM 500mg."

text = text.lower()

text = re.sub(r'\bpcm\b', 'paracetamol', text)
text = re.sub(r'\bbp\b', 'blood pressure', text)

tokens = re.findall(r'\d+/\d+|\d+(?:\.\d+)?\s?mg|[a-z]+', text)

print("Normalized Text:", text)
print("Tokens:", tokens)
```

```python
def validate_patient_id(value):
    states = {
        0: {'P': 1},
        1: {'1': 2, '2': 2, '3': 2, '4': 2,
            '5': 2, '6': 2, '7': 2, '8': 2, '9': 2, '0': 2}
    }

    state = 0

    for char in value:
        if state == 0 and char == 'P':
            state = 1
        elif state == 1 and char.isdigit():
            state = 2
        elif state == 2 and char.isdigit():
            state = 2
        else:
            return False

    return state == 2 and len(value) == 6

print(validate_patient_id("P12345"))
print(validate_patient_id("A12345"))
```
